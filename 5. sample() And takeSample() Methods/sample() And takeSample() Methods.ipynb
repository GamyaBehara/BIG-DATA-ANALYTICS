```python
from pyspark.sql import SparkSession

# Step 1: Initialize Spark Session
spark = SparkSession.builder.appName("SamplingExample").getOrCreate()
```

```python
# Step 2: Read CSV file into DataFrame
df = spark.read.csv("students.csv", header=True, inferSchema=True)
```

```python
# === Sampling Demonstration (within 7 operations) ===

# 1. View first 5 rows
print("=== First 5 rows of dataset ===")
df.show(5)
```

```python
# 2. Print schema
print("=== Schema of dataset ===")
df.printSchema()
```

```python
# 3. Random sample without replacement (30% of data)
print("=== Sample (30% without replacement) ===")
df.sample(withReplacement=False, fraction=0.3, seed=42).show(10)
```

```python
# 4. Random sample with replacement (20% of data)
print("=== Sample (20% with replacement) ===")
df.sample(withReplacement=True, fraction=0.2, seed=42).show(10)
```

```python
# 5. Take a random sample of 5 rows using takeSample (without replacement)
print("=== takeSample: 5 rows (without replacement) ===")
sampled_rows = df.rdd.takeSample(False, 5, seed=42)
for row in sampled_rows:
    print(row)
```

```python
# 6. Take a random sample of 5 rows using takeSample (with replacement)
print("=== takeSample: 5 rows (with replacement) ===")
sampled_rows_wr = df.rdd.takeSample(True, 5, seed=42)
for row in sampled_rows_wr:
    print(row)
```

```python
# 7. Count total rows (to compare with sampled data size)
print("Total rows in dataset:", df.count())
```

# Summary

This experiment demonstrates how to perform **sampling operations in PySpark** using a student dataset stored as a CSV file. The dataset contains 50 records with columns for ID, name, age, gender, and marks in three subjects (Math, Science, English).  

## Operations Performed

1. **Data Loading**
   - Loaded `students.csv` into a PySpark DataFrame using `spark.read.csv()`.  
   - Verified dataset with `.show()` and inspected schema using `.printSchema()`.  

2. **Random Sampling**
   - `sample(withReplacement=False, fraction=0.3)` → Selected ~30% of records without replacement.  
   - `sample(withReplacement=True, fraction=0.2)` → Selected ~20% of records with replacement.  

3. **takeSample Operation**
   - `takeSample(False, 5)` → Retrieved 5 random rows without replacement.  
   - `takeSample(True, 5)` → Retrieved 5 random rows with replacement.  

4. **Row Count**
   - Verified the dataset size using `.count()` which returned **50 records**.  

## Key Insights
- PySpark provides flexible methods for sampling data, both with and without replacement.  
- `sample()` is useful for creating subsets of large datasets.  
- `takeSample()` directly returns a small sample as Python objects for quick inspection.  
- Sampling is an essential step for exploratory analysis, testing, and model prototyping on big data.
